```python
import os
import shutil
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
path = r"C:\Users\user\OneDrive - REVA University\Projects\PlantHealth_AI\Dataset\PlantVillage"
test_path = r"C:\Users\user\OneDrive - REVA University\Projects\PlantHealth_AI\PlantVillage\test"
train_path = r"C:\Users\user\OneDrive - REVA University\Projects\PlantHealth_AI\PlantVillage\train"
validation_path = r"C:\Users\user\OneDrive - REVA University\Projects\PlantHealth_AI\PlantVillage\Validation"
```

```python
folders = os.listdir(path)

classes = []

for folder in folders:
    full_path = os.path.join(path, folder)

    if os.path.isdir(full_path):
        print(folder)
        classes.append(folder)
```

```python
for cls in classes:
    os.makedirs(os.path.join(test_path, cls), exist_ok=True)
    os.makedirs(os.path.join(train_path, cls), exist_ok=True)
    os.makedirs(os.path.join(validation_path, cls), exist_ok=True)
```

```python
for class_name in os.listdir(path):

    class_path = os.path.join(path, class_name)

    if not os.path.isdir(class_path):
        continue

    images = [
        file for file in os.listdir(class_path)
        if file.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    random.shuffle(images)

    total = len(images)

    train_count = int(total * 0.70)
    val_count = int(total * 0.15)

    train_images = images[:train_count]
    val_images = images[train_count:train_count + val_count]
    test_images = images[train_count + val_count:]

    os.makedirs(os.path.join(train_path, class_name), exist_ok=True)
    os.makedirs(os.path.join(validation_path, class_name), exist_ok=True)
    os.makedirs(os.path.join(test_path, class_name), exist_ok=True)

    for image in train_images:
        shutil.copy2(
            os.path.join(class_path, image),
            os.path.join(train_path, class_name, image)
        )

    for image in val_images:
        shutil.copy2(
            os.path.join(class_path, image),
            os.path.join(validation_path, class_name, image)
        )

    for image in test_images:
        shutil.copy2(
            os.path.join(class_path, image),
            os.path.join(test_path, class_name, image)
        )

    print(
        f"{class_name}: "
        f"Train={len(train_images)}, "
        f"Validation={len(val_images)}, "
        f"Test={len(test_images)}"
    )

print("\nDataset splitting completed!")
```
